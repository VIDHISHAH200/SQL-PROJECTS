#  1) find profit margin by category  and sub categroy


```python
# extracting  new column month and year from order date
```

```python
v['year']=v['Order Date'].dt.year
v['month']=v['Order Date'].dt.month
```

# 1) Which product categories and sub-categories generate the highest and lowest profit margins?

```python
catprofit=v.groupby(['Category','Sub-Category'])[['Sales','Profit']].sum().reset_index()
print("\n1. Profit margins by category and sub-category")
print(catprofit)
```

# Which states or regions contribute most to overall sales and profits?


```python
profit = v.groupby(['State','Region'])[['Sales', 'Profit']].sum().reset_index()
print("\n2. Profit by state and regions")
print(profit)
```

```python
profit
```

# What is the impact of discounts on profit across different product categories?


```python
avg=v.groupby('Discount')[['Sales','Profit']].mean().reset_index()
print("/n3. Average impact of discount on profit")
print(avg)
```

# 4) are there any products that consistently generate losses despite good sales volume?


```python
loss_products =v.groupby('Product Name')[['Sales','Profit']].sum().sort_values(by='Profit').head(10).reset_index()
print("/n4.Loss making products despite good sales")
print(loss_products)
```

# operational and shipping analysis

# 1) What is the average delivery time (Ship Date - Order Date) by Ship Mode, and how does it impact customer segment satisfaction?


```python
avgshipmode=v.groupby('Ship Mode')['Delivery time'].mean().reset_index()
print("/n5. Average delivery time by ship mode")
print(avgshipmode)
```

 # 6 Which shipping modes are most frequently used, and how do they correlate with profit or loss?

```python
shipmode=v.groupby('Ship Mode')[['Sales','Profit']].sum().reset_index()
print("\n.6 Most frequently used shipping models")
print(shipmode)
```


 #7) do sales and profits differ by shipping mode across regions?

```python
ap=v.groupby(['Ship Mode','Region'])[['Sales','Profit']].sum().reset_index()
print("/n7. Shipping performance by region")
print(ap)
```

#8 Which customer segments (Consumer, Corporate, Home Office) have the highest average order value?

```python
from numpy._core.defchararray import count
aov = v.groupby('Segment').agg({'Sales': 'mean','Order ID':'count'})
print("\8. Customer segment with highest average order value")
print(aov)
```

# 9 Top 10 profitable customerS

```python
psc=v.groupby(['Customer ID','Customer Name'])[['Profit']].sum().sort_values(by ='Profit',ascending=False).reset_index().head(10)
print("\n9. TOP 10 profitable customers ")
print(psc)
```

# 10  10. Repeat customers and discount/profit

```python
rc=v.groupby('Customer ID').agg({'Order ID':'nunique','Profit':'sum','Discount':'mean'}).reset_index()
print("\10. Repeat customer behaviour")
print(rc)
```

# 13. Sub-category sales per unit


```python
value_per_unit = v.groupby('Sub-Category').agg({'Sales':'sum','Quantity':'sum'})
value_per_unit['Sales per unit']=value_per_unit['Sales']/value_per_unit['Quantity']
value_per_unit= value_per_unit.reset_index()
value_per_unit
```

# 14. Product combinations (market basket)


```python
market=v.groupby('Order ID')['Product Name'].apply(lambda x:list(x)).reset_index
print("\n14. Product combinations")
print(market)
```

# 15. Predicting profit using regression


```python
x =v[['Sales','Quantity','Discount']]
y =v['Profit']
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=0)
model = LinearRegression().fit(xtrain,ytrain)
model_score =model.score(xtest,ytest)
model_score
print(f"\n15. Predictive model estimation:{model_score:.4f}")
```
